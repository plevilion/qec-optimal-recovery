```python
import numpy as np
import matplotlib.pyplot as plt

from robustness import mismatched_petz_fidelity
```

```python
gamma_0 = np.linspace(0, 0.3, 31)   
m_5 = [mismatched_petz_fidelity(0.05, gam) for gam in gamma_0]  
m_1 = [mismatched_petz_fidelity(0.1, gam) for gam in gamma_0]            
m_2 = [mismatched_petz_fidelity(0.2, gam) for gam in gamma_0]            


plt.plot(gamma_0, m_5, label=r"$\gamma = 0.05$")
plt.plot(gamma_0, m_1, label=r"$\gamma = 0.1$")
plt.plot(gamma_0, m_2, label=r"$\gamma = 0.2$")
plt.plot(0.05, mismatched_petz_fidelity(0.05, 0.05), "o", color="C0")
plt.plot(0.1,  mismatched_petz_fidelity(0.1, 0.1),   "o", color="C1")
plt.plot(0.2,  mismatched_petz_fidelity(0.2, 0.2),   "o", color="C2", label=r"$\gamma_0 = \gamma$")
plt.xlabel(r"assumed gamma $\gamma_0$")
plt.ylabel(r"mismatched entanglement fidelity $F(\gamma, \gamma_0)$")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig("../figures/mismatched_fidelity.png", dpi=200, bbox_inches="tight")
plt.show()
```

Remark : at $\gamma_0 = 0$, the Petz recovery map with parameter $\gamma_0$ is just the baseline recovery implemented in recovery.py (recovery by $V^{\dagger}$) and the values for $F(\gamma, 0)$ match the ones found in the petz_vs_baseline notebook. 

First interpretation : the maximum of $F(\gamma, \gamma_0)$ is not attained at $\gamma_0 = \gamma$ implying that there are potentially Petz map constructed on incorrect noise parameters that could outperform the Petz map for the true noise.
